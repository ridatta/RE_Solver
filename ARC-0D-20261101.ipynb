## Instantiate solver and define settings

```python
solver = RE.solver0D()
solver.display() # Show default values
```

```python
# Change solver settings
solver.T0 = 3e3 # Initial temperature [eV]
solver.Tf = 5 # final temperature [eV]
solver.t0 = 0.1e-3
solver.R0 = 4.62
solver.a = 1.18
solver.k = 1.515 # Elongation
solver.ni = 2.2e20
solver.I0 = 11.9e6
solver.Btor = 11.4
solver.L = 3.18e-6
# Sources
solver.Ir0 = 1.0e3 # Initial RE Current [A]
solver.iCompton = 0 # Turn on activated sources
solver.iTritium = 0
solver.iAvalanche = 3 # Hesslow
solver.iDreicer = 0

# Impurities (approx. baseline MGI of Ne + D2; full assimilation)
solver.n_imp = 7.28e19
solver.nD = 18 * solver.n_imp

# Other
solver.tstart = 0.8e-3 # ms
solver.Zeff_scale = 0.9
solver.device='ARC'

solver.display()

Ap = np.pi*solver.a**2*solver.k
Vol = Ap * (2*np.pi*solver.R0)

print('Mag energy = %1.3g MJ'%(0.5*solver.I0**2*solver.L/1e6))
print('Plasma Area = %1.3f m2'%(np.pi*solver.a**2*solver.k))
print('Injected Ne atoms = %1.3g'%(Vol * solver.n_imp))
print('Injected D2 molecules = %1.3g'%(Vol * solver.nD/2))
```

## Solve

```python
# solve
solver.tmax = 10e-3 # end time [s]
solver.idiag = True # calculate detailed output
t, Ip, Ir, Wk = solver.solve()
```

Here, we will plot  𝐼𝑝 and  𝐼_𝑅𝐸 as a function of time

```python
fig, axs = plt.subplots(1,1,figsize=(3,2),dpi=200,sharex=True)
plt.plot(t*1e3,Ip/1e6,label='$I_p$')
plt.plot(t*1e3,Ir/1e6,label='$I_r$')
plt.grid()
plt.xlabel('t [ms]')
plt.ylabel('I [MA]')
plt.ylim([0,12])
plt.yticks(np.arange(0,13,2))
plt.xlim([0,solver.tmax*1e3])

plt.legend()
```

We can also plot the RE beam energy as a function of time

```python
fig, axs = plt.subplots(1,1,figsize=(3,2),dpi=200,sharex=True)

plt.plot(t*1e3,Wk/1e6)
plt.grid()
plt.xlabel('t [ms]')
plt.ylim(ymin=0)
plt.xlim([0,solver.tmax*1e3])
axs.set_ylabel('$W_{kin}$ [MJ]')


print('Plateau Energy [MJ] = %1.3f'%(Wk[-1]/1e6))

# Estimate the average RE energy in MeV
N = Ir[-1] * (2 * np.pi * solver.R0) / (const.e * const.c)
avg_energy = Wk[-1] / N
print('Average RE Energy [MeV] = %1.3f'%(avg_energy/1e6/const.eV))
```

# Hot-Tail Seed

Shows that for the m3dc1 run, the expected HT seed in negligible

We use volume averaged 0D parameters

```python

# Temp evolution
data = np.loadtxt('data/t_Te_m3dc1_ARC-06.csv')
tp = data[156:-1,0]
Tp = data[156:-1,1] + 0
tp = tp - tp[0]

# Re-sample
tgrid = np.linspace(tp[0],tp[-1],5000)
Tp = np.interp(tgrid,tp,Tp)
tp = np.interp(tgrid,tp,tp)



plt.plot(tp*1e3,Tp)
plt.yscale('log')
plt.xlim([0,2])

# Other Parameters

eta0 = 0.51 * getTransSpitzerResistivity(15,Tp[0],1.28) * 0.01 # ohm-m
Ap = np.pi*solver.a**2*solver.k
E0 = eta0 * solver.I0 / Ap # V/m, Initial Electric Field
ne0 = 2e20 #d['ne'][0] # Electron density per cubic m

print('E0 [V/m] = %1.3f'%E0)
print('ne0 [m-3] = %1.3g'%ne0)
```

```python

# Initiate arrays
nht = np.zeros_like(tp)
nht_rate = np.zeros_like(tp)
dt = np.diff(tp)[0]

# Caluclate
for ii in range(1,tp.shape[0]):
    tnow = tp[ii]
    T = Tp[ii] # eV
    nht_rate[ii], nht[ii] = RE.get_HT_rate(ne=ne0, # electron density
                                           T=T, # temperature
                                           t=tnow, # time
                                           T0=Tp[0], # initial temperature
                                           E0=E0, # initial E-field
                                           nht=nht[ii-1],
                                           dt=dt)
```

```python
# Plot
fig, axs = plt.subplots(3,1,sharex=True,figsize=(2,5),dpi=200)
axs[0].plot(tp*1e3,nht/ne0,'-')
axs[0].set_ylabel('$n_{HT}/n_0$ [-]')
axs[0].set_yscale('log')
# axs[0].set_ylim([1e-20,1e0])
axs[0].grid()

axs[1].plot(tp*1e3,nht_rate / 1e3,'-')
axs[1].set_ylabel('HT Rate [$m^{-3} ms^{-1}$]')
axs[1].set_ylim(ymin=0)
axs[1].grid()


axs[2].plot(tp*1e3,Tp)
axs[2].set_ylabel('T [eV]')
axs[2].set_ylim(ymin=1)
axs[2].grid()
axs[2].set_yscale('log')


axs[-1].set_xlim([0,2])
axs[-1].set_xlabel('t [ms]')


# HT Current
Iht = nht[-1] * const.c * const.e * np.pi * solver.a**2 * solver.k
print('Iht = %1.3g kA'%(Iht/1e3))


axs[0].set_ylim([1e-20,1e-14])
```
